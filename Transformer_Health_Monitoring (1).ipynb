```python
import pandas as pd
import matplotlib.pyplot as plt

df2=pd.read_excel("Transformer_Maintenance_History_2")
df=pd.read_excel("Transformer_Operational_Loading_2")

def data_read(df,df2):
    print("=======Maintenance_history File info========")
    display(df.describe())
    display(df.head())
    print("=======Operational_Loading info=====")
    display(df2.describe())
    display(df2.head())

def data_inspection(df,df2):
    print("total enteries in the file 1 ==", len(df))
    print("total  Transformer enteries  in file 1 =",df["Transformer_ID"].count())
    null_count=df["Transformer_ID"].isna().sum()
    
    if null_count !=0:
        print("Any rows with the null transformer id ==",null_count)
        display("Those enteries are = ", df[df["Transformer_ID"].isna()])
    else:
        print("NO missing Transformer ID fount")

    empty_count= df.isna().all(axis=1).sum()
    if empty_count !=0:
        print(" Number of the completely empty rows = =",empty_count)
        display("those rows are =",df[df.isna().all(axis=1)])
        clean_df= df.dropna(how="all")
        print("clean dataframe has been create with total enteries",len(clean_df))
        print("unique transformer id",clean_df["Transformer_ID"].nunique())

    else:
        clean_df=df.copy()
        print("No completely empty rows")


    print("total number of the enteries in the file2 == ", len(df2))
    empty_count2= df2.isna().all(axis=1).sum()
    if empty_count2 != 0:
        print("any completely empty rows ==",empty_count2)
        display("Those rows are as=",df2[df2.isna().all(axis=1)])
        clean_data2=df2.dropna(how="all")
        print("new cleaned data frame 2 has been created with total no of the enteries =",len(clean_data2))
        print(" if any missing maiteneance id ",clean_data2["Transformer_ID"].isna().sum())

    else:
        clean_data2= df2.copy()
        print("no completely empty data in the file 2")

    duplicate_enteries2= clean_data2["Transformer_ID"].duplicated() & clean_data2["Transformer_ID"].notna()
    if duplicate_enteries2.sum() !=0:
        print("there are duplicates enteries also =", duplicate_enteries2.sum())
        display("those entereis are =", clean_data2[clean_data2["Transformer_ID"].duplicated()& clean_data2["Transformer_ID"].notna()])
        print(" keeping the latest record")
        latest_recored2=clean_data2.sort_values(by="Maintenance_Date",ascending= True).drop_duplicates(subset=["Transformer_ID"],keep="last")
        print(len(latest_recored2))
        print(" combined both reports")
        merged_report=pd.merge(clean_df,latest_recored2,on="Transformer_ID",how="left")
        display(merged_report)

    else:
        merged_report=pd.merged(clean_df,latest_recored2,on="Transformer_ID",how="left")
        print(" No duplicates have been found , here is the merged data",merged_report)

    return merged_report


def risk_def(merged_report):
    risk_level=merged_report.copy()
    Warning=(merged_report["Loading_%"]>=100) ^ (merged_report["Oil_Temp_C"]>=80)
    Critical=(merged_report["Loading_%"]>=100 ) & (merged_report["Oil_Temp_C"]>=80)
    Normal=( merged_report["Loading_%"]<100 ) & ( merged_report["Oil_Temp_C"]<80 )
    Unassigned=(merged_report["Loading_%"].isna()) | (merged_report["Oil_Temp_C"].isna())
    risk_level.loc[Warning,"Risk_level"]= "Warning"
    risk_level.loc[Critical,"Risk_level"]="Critical"
    risk_level.loc[Normal,"Risk_level"]= "Normal"
    risk_level.loc[Unassigned,"Risk_level"]= "Unassigned"
    display(risk_level)
    return risk_level



def graphs_Plot(risk_level):


    plt.figure(figsize=(18,6))    
    top_20= risk_level.sort_values(by="Loading_%", ascending=False).head(20)
    LIMIT=100
    plt.axhline(y=LIMIT,color="red")
    plt.bar(top_20["Transformer_ID"],top_20["Loading_%"])
    plt.title("Top 20 loading Transformer vs Loading")
    plt.xlabel("Transformer_ID")
    plt.ylabel("Loading_%")
    plt.grid(True)
    plt.show()


    plt.figure(figsize=(12,4))
    riskcount=risk_level["Risk_level"].value_counts()
    plt.bar(riskcount.index,riskcount)
    plt.grid(True)
    plt.show()
    print("Total risk count values",riskcount)


    
    
    



data_read(df,df2)
merged_report = data_inspection(df, df2)
risk_level=risk_def(merged_report)
graphs_Plot(risk_level)
    
```
