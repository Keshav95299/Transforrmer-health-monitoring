```python
import pandas as pd
import matplotlib.pyplot as plt
df= pd.read_excel("data/transformer_loading_final.xlsx")
df1= pd.read_excel("data/maintenance_history_final.xlsx")

def data_reading(df,df1):
    print("====Transformer Loading Data====")
    display(df.describe())
    display(df.head())
    print("====Maintenece History Data===")
    display(df1.head())
    display(df1.describe())
    display("==Any missing data from the datset===\n",df.isna().sum())
    display("=== Those data are===",df[df.isna().any(axis=1)])

def statistical_information(df):
    
    print("==Overloading Transformers ==\n", df.loc[df["Loading_%"]>100,["Transformer_ID","Substation","Loading_%"]].to_string(index=False))
    print("== Higher Oil temp==\n",df.loc[df["Oil_Temp_C"]>80,["Transformer_ID","Substation","Oil_Temp_C"]].to_string(index=False))
    print("==statical information about transformer===\n",df["Loading_%"].agg(["mean","max","min"]))
    print("==statical information about the OIl===\n",df["Oil_Temp_C"].agg(["mean","max","min"]))

def cobmined_transformer_information(df,df1):
    combined=pd.merge(df,df1,on="Transformer_ID",how="left")
    display(combined[["Transformer_ID","Loading_%","Substation","Maintenance_Type","Maintenance_Date"]])
    return combined

   
    

def risk_level(df):
    print("Risk level==")
    warning=  (df["Loading_%"]>100 ) ^ (df["Oil_Temp_C"]>80)
    critical= (df["Loading_%"]>100 ) & (df["Oil_Temp_C"]>80)
    normal= (df["Loading_%"]<=100 )  & (df["Oil_Temp_C"]<=80)
    missing_data= (df["Loading_%"].isna() ) | (df["Oil_Temp_C"].isna())
    Risky=df.copy()
    Risky.loc[warning,"Risk level"]="Warning"
    Risky.loc[critical,"Risk level"]="Critical"
    Risky.loc[normal,"Risk level"]="Normal"
    Risky.loc[missing_data,"Risk level"]="Unknown"
    display(Risky[["Transformer_ID","Loading_%","Risk level"]])
    print("total missing rows to the Risky\n",Risky["Risk level"].isna().sum())
    print(" data where missing assign\n",Risky.loc[Risky["Risk level"].isna(),["Transformer_ID","Loading_%","Oil_Temp_C","Risk level"]])
    return Risky





def Graphs(df,Risk_file):
    plt.figure(figsize=(18,6))
    top20 = df.sort_values(by="Loading_%", ascending=False).head(20)
    LIMIT=100
    plt.axhline(y=LIMIT,color="red")
    plt.bar(top20["Transformer_ID"],top20["Loading_%"])
    plt.xlabel("Transformer_ID")
    plt.ylabel("Loading_%")
    plt.title(" Top 20 -Transformer vs Loading")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8,5))
    plt.title("Risk level vs count")
    risk_count=Risk_file["Risk level"].value_counts()
    plt.bar(risk_count.index,risk_count.values)
    plt.grid(True)
    plt.show()
    print("total risk count",Risk_file["Risk level"].value_counts())

    

            

data_reading(df,df1)
statistical_information(df)
cobmined_transformer_information(df,df1)
Risk_file = risk_level(df)
Graphs(df,Risk_file)
```
